```python
import spacy
from spacy import displacy

text = 'I give a lot of importance to kindness, care and affection, hence health care centers like Apollo,Cradle are my preferred choices'
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
```

```python
for ent in doc.ents:
  # print(ent)
  print(ent.text, ent.label_)
```

```python
print(f"{'Entity':<20} | {'Label':<10} | {'Description'}")
print("-" * 50)
for ent in doc.ents:
    print(f"{ent.text:<20} | {ent.label_:<10} | {spacy.explain(ent.label_)}")
```

```python
displacy.render(doc, style="ent", jupyter=True)
```
