```python
!pip install spacy
!python -m spacy download en_core_web_sm
import spacy as sp
nlp = sp.load('en_core_web_sm')
```

```python
sentence1 = nlp('I will achieve what I have aim to achieve and I will succeed')

print('sentence 1')
print('-----')
for token in sentence1:
  #  print(token)
  #  print('-----')
  #  print(token.text, token.pos_)
   print(f"{token.text:10} {token.pos_}")
  #  print(token.text, token.pos_, token.dep_, token.lemma_)
```

```python
print('sentence 2')
print('-----')
sentence2 = nlp("Please follow our blog to see more information.")
for token in sentence2:
  print(f"{token.text:10} {token.pos_}")
```

```python
from spacy import displacy
displacy.serve(sentence1, style='dep')
```

```python
displacy.serve(sentence2, style='dep')
```
