```python
#imports
from flair.data import Sentence
from flair.models import SequenceTagger
```

```python
# load the POS tagger
tagger = SequenceTagger.load('pos')

#create text data
text = "Flair is a great library for Natural language processing, you must check it out."
sentence = Sentence(text) # text preprocessing


```

```python
# Predictions
tagger.predict(sentence)
sentence.to_tagged_string()
```
